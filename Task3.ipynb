```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"D:/data/zepto.csv")
```

```python
plt.figure(figsize=(12,6))
df['Category'].value_counts().plot(kind='bar')
plt.title("Products by Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
top10 = df.sort_values(
    by='mrp',
    ascending=False
).head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    data=top10,
    x='mrp',
    y='name'
)

plt.title("Top 5 Expensive Products")
plt.show()
#2. Top 5 most expencive product
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['discountPercent'],
    bins=20
)

plt.title("Discount Percentage Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='mrp',
    y='discountedSellingPrice'
)

plt.title("MRP vs Discounted Price")
plt.show()
```

```python
df['outOfStock'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Out Of Stock Products")
plt.ylabel("")
plt.show()
```

```python
avg_price = df.groupby(
    'Category'
)['mrp'].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6))

avg_price.plot(kind='bar')

plt.title("Average MRP by Category")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['weightInGms'],
    bins=30
)

plt.title("Weight Distribution")
plt.show()
```

```python
avg_discount = df.groupby('Category')['discountPercent'].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6))
avg_discount.plot(kind='bar')

plt.title("Average Discount by Category")
plt.xlabel("Category")
plt.ylabel("Average Discount (%)")
plt.xticks(rotation=45)

plt.show()
```

```python
print("Total Products :", len(df))
print("Total Categories :", df['Category'].nunique())
print("Average Discount :", round(df['discountPercent'].mean(),2))
print("Average MRP :", round(df['mrp'].mean(),2))
```
